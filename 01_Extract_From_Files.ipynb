```python
# for Access and write Data from SQL DB
import pandas as pd
import os
import pyodbc
```

```python
pyodbc.drivers()
```

```python
Connection_string = (
    r"Driver={ODBC Driver 17 for SQL Server};"
    r"Server=ASUS\SQLEXPRESS;"
    r"Database=Test;"
    r"Trusted_Connection=yes;"

)

conn = pyodbc.connect(Connection_string)
print("Connection is succsful")

```

```python

cursor=conn.cursor()
cursor.execute("select * from dbo.orders")

for row in cursor.fetchall():
  print(row)
```

```python
quary = """
select * 
from 
dbo.Orders """
Orders=pd.read_sql(quary , conn)
Orders.head(100)
```

```python
quary="""
select * from [Test].[dbo].[orders]
where YEAR(shipped_date)=2016
;
"""

Orders=pd.read_sql(quary , conn)
print(Orders.count());
```
